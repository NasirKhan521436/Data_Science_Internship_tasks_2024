```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
sns.set(color_codes=True)
%matplotlib inline 
import warnings
warnings.filterwarnings("ignore")

file_path = r"E:\COLLEGGE\6 SEMESTER\Internship cognifyz\Dataset .csv"


try:
     dataset = pd.read_csv(file_path, sep=",", on_bad_lines='skip', encoding="utf-8-sig")
     print("FILE READ SUCESSFULLY")
except FileNotFoundError:
     print(f"File not found: {file_path}")
```

 Extract additional features from the existing
 columns, such as the length of the restaurant
 name or address.

```python
dataset["Address Length"]=dataset["Address"].apply(lambda x: len(str(x)))
dataset["Restaurant Name Length"]=dataset["Restaurant Name"].apply(lambda x: len(str(x)))

dataset[["Restaurant Name","Restaurant Name Length","Address","Address Length"]]
```

Create new features like "Has Table Booking"
 or "Has Online Delivery" by encoding
 categorical variables

```python
#Feature Engineering Part II:Create new features like "Has Table Booking"or "Has Online Delivery" by encoding
 #categorical variables.
dataset["Categ_Has Table booking"]=dataset["Has Table booking"].apply(lambda x: 1 if x=="Yes" else 0)
dataset["Categ_Has Online delivery"]=dataset["Has Online delivery"].apply(lambda x: 1 if x=="Yes" else 0)

dataset[["Has Table booking","Categ_Has Table booking","Has Online delivery","Categ_Has Table booking"]]

  
```
