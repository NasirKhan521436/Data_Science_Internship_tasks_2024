```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
sns.set(color_codes=True)
%matplotlib inline 
import warnings
warnings.filterwarnings("ignore")
```

```python
file_path = r"E:\COLLEGGE\6 SEMESTER\Internship cognifyz\Dataset .csv"

try:
     dataset = pd.read_csv(file_path, sep=",", on_bad_lines='skip', encoding="utf-8-sig")
     print("File read sucessfully")
except FileNotFoundError:
     print(f"File not found: {file_path}")
```

 Create visualizations to represent the distribution
 of ratings using different charts (histogram, bar
 plot, etc.)

```python
plt.figure(figsize=(10, 6))
sns.countplot(x=dataset["Aggregate rating"], data=dataset , color="pink")
plt.title("COUNT OF  RATINGS")
plt.xlabel("Rating------->")
plt.ylabel("Count ------->")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=dataset["Aggregate rating"], data=dataset , color="Blue")
plt.title("Distribution of Ratings")
plt.xlabel("Ratings------->")
plt.ylabel("Count ------->")
plt.show()
```

Compare the average ratings of different cuisines
 or cities using appropriate visualizations.

```python
average_rating_by_city = dataset.groupby("City")["Aggregate rating"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=average_rating_by_city.head(10).index, y=average_rating_by_city.head(10).values, palette="viridis")
plt.title("Average Ratings of Different Cities (Top 10)")
plt.xlabel("City")
plt.ylabel("Average Ratings")
plt.xticks(rotation=45)
plt.show()
```
