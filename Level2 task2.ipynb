```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
sns.set(color_codes=True)
%matplotlib inline 
import warnings
warnings.filterwarnings("ignore")

file_path = r"E:\COLLEGGE\6 SEMESTER\Internship cognifyz\Dataset .csv"


try:
     dataset = pd.read_csv(file_path, sep=",", on_bad_lines='skip', encoding="utf-8-sig")
     print("FILE READ SUCESSFULLY")
except FileNotFoundError:
     print(f"File not found: {file_path}")
```

Determine the most common price range
 among all the restaurants.

```python
# Determine the most common price rangeamong all the restaurants.
dataset["Price range"].value_counts()
```

```python
#using mode for calculating most common 
most_common=dataset["Price range"].mode()[0]
print("The most common price range among all restaurants is :",most_common)
```

 Calculate the average rating for each price
 range.

```python
#.Calculate the average rating for each pricerange.Identify the color that represents the highestaverage rating among different price ranges.
#Part I:
Average_rating_Price_range=dataset.groupby("Price range")["Aggregate rating"].mean().round(2)
print("Average rating for each price range is :",Average_rating_Price_range)

```

 Identify the color that represents the highest
 average rating among different price ranges

```python
#Part II: Identify the color that represents the highestaverage rating among different price ranges
Highest_avg_rating=Average_rating_Price_range.idxmax()
plt.bar(Average_rating_Price_range.index, Average_rating_Price_range, color="Orange" ,width=0.6)

plt.bar(Highest_avg_rating, Average_rating_Price_range[Highest_avg_rating], color="green" ,width=0.6)
plt.title("Average Rating by Price Range")
plt.xlabel("Price Range---->")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```
