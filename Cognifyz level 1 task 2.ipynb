```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
sns.set(color_codes=True)
%matplotlib inline 
import warnings
warnings.filterwarnings("ignore")
```

```python
file_path = r"E:\COLLEGGE\6 SEMESTER\Internship cognifyz\Dataset .csv"

try:
     dataset = pd.read_csv(file_path, sep=",", on_bad_lines='skip', encoding="utf-8-sig")
     print("File read sucessfully")
except FileNotFoundError:
     print(f"File not found: {file_path}")
```

 Explore the distribution of categorical
 variables like "Country Code," "City," and
 "Cuisines."

```python
#Explore the distribution of categorical varriables like country code city , and cyuisenss  identify the top cuiness and cities with the highest number of  resaurants 
plt.figure(figsize=(10,7))
sns.countplot(x="Country Code",palette="Set1" ,data=dataset )
plt.title("Distribution of Resaurants by Country Codes  ")
plt.xlabel("country code  ")
plt.ylabel("Number of Restaurants  ")
plt.show()
```

```python
top_countries=dataset["Country Code"].value_counts().head(7)
print("the top countries are ", top_countries)
```

```python
plt.figure(figsize=(10,7))
sns.countplot(x="City",palette="Set1" ,data=dataset , order=dataset["City"].value_counts().head(20).index)
plt.title("Distribution of Resaurants by City ")
plt.xlabel("City  ")
plt.ylabel("Number of Restaurants  ")
plt.xticks(rotation=45)
plt.show()
```

```python
# Create a pie chart for representing
city_counts = dataset['City'].value_counts().head(5)


plt.figure(figsize=(8, 8))
plt.pie(city_counts, labels=city_counts.index, autopct='%1.1f%%', colors=sns.color_palette("Set1"))
plt.title("Distribution of Restaurants by top 5  Cities")
plt.show()
```

```python
plt.figure(figsize=(16,6))
cou_count=dataset['Cuisines'].value_counts()
cou_count.head(20).plot(kind="bar", color=sns.color_palette("Set1"))

plt.title("Distribution of Resaurants by  top 20 highest number of Cuisines ")
plt.xlabel("Number of Cuisines  ")
plt.ylabel("Number of Restaurants  ")
plt.xticks(rotation=20)
plt.show()
```

```python

cou_count = dataset['Cuisines'].value_counts()
plt.figure(figsize=(16, 6))
cou_count.head(20).plot(kind="hist", color=sns.color_palette("Set2"))

plt.title("Distribution of Restaurants by the Top 20 Highest Number of Cuisines")
plt.xlabel("Number of Cuisines")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=20)
plt.show()
```
